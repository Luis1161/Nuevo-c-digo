###Descripción
Una empresa constructora ha sido contratada para diseñar una piscina con forma de paralelepípedo rectangular. La piscina debe contener exactamente $48 ~\text{m}^3$ de agua. Además, por razones de espacio en el terreno, el largo de la piscina debe ser el triple que el ancho.

La empresa constructora quiere determinar las dimensiones (ancho, $a$, largo, $l$ y profundidad, $p$), que minimizan la superficie total y, en consecuencia, la cantidad de material necesario para recubrir las paredes y el fondo de la piscina.


**a) Obtén una expresión matemática para la cantidad de material, $C$ (en $\text{m}^2$), en función del ancho, $a$ (en $\text{m}$), de la piscina.**

**Solución:**
La cantidad de material necesario tanto para cubrir el piso de junto a las partes frontales y laterales de la piscina se obtienen de la siguiente fórmula:

$$C=al+2ap+2lp$$

El largo debe ser el triple del ancho por ende:

$$l=3a$$

Se tiene un volumen fijo, de $48 \text{m}^3$

$$V=alp $$

Reemplazando los valores:

$$48=alp$$

Se despeja $p$:

$$ p=\dfrac{48}{al}$$

Y se sustituye $l=3a$:

$$p=\dfrac{48}{a(3a)}$$

Ahora se simplifica:

$$p=\dfrac{48}{3a^2}=\dfrac{16}{a^2}$$

Al reemplazar entonces el valor de $C$ queda:

$$C=a(3a)+2a\dfrac{16}{a^2}+2(3a)(\dfrac{16}{a^2})$$

$$C(a)=3a^2+\dfrac{32}{a}+\dfrac{96}{a}$$

En presencia de fracciones homogéneas se puede simplificar un poco más:

$$C(a)=3a^2+\dfrac{128}{a}$$
