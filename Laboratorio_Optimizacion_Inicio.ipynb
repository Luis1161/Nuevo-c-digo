###Descripción
Una empresa constructora ha sido contratada para diseñar una piscina con forma de paralelepípedo rectangular. La piscina debe contener exactamente $48 ~\text{m}^3$ de agua. Además, por razones de espacio en el terreno, el largo de la piscina debe ser el triple que el ancho.

La empresa constructora quiere determinar las dimensiones (ancho, $a$, largo, $l$ y profundidad, $p$), que minimizan la superficie total y, en consecuencia, la cantidad de material necesario para recubrir las paredes y el fondo de la piscina.


**a) Obtén una expresión matemática para la cantidad de material, $C$ (en $\text{m}^2$), en función del ancho, $a$ (en $\text{m}$), de la piscina.**

**Solución:**
La cantidad de material necesario tanto para cubrir el piso de junto a las partes frontales y laterales de la piscina se obtienen de la siguiente fórmula:

$$C=al+2ap+2lp$$

El largo debe ser el triple del ancho por ende:

$$l=3a$$

Se tiene un volumen fijo, de $48 \text{m}^3$

$$V=alp $$

Reemplazando los valores:

$$48=alp$$

Se despeja $p$:

$$ p=\dfrac{48}{al}$$

Y se sustituye $l=3a$:

$$p=\dfrac{48}{a(3a)}$$

Ahora se simplifica:

$$p=\dfrac{48}{3a^2}=\dfrac{16}{a^2}$$

Al reemplazar entonces el valor de $C$ queda:

$$C=a(3a)+2a\dfrac{16}{a^2}+2(3a)(\dfrac{16}{a^2})$$

$$C(a)=3a^2+\dfrac{32}{a}+\dfrac{96}{a}$$

En presencia de fracciones homogéneas se puede simplificar un poco más:

$$C(a)=3a^2+\dfrac{128}{a}$$


**b) Calcula la primera derivada de la función en forma manual. Registra el procedimiento paso a paso.**

**Solución:**

Para calcular la derivada se tiene que expresar la fracción en el término donde se pueda aplicar cabalmente la regla de la potencia:

$$C(a)=3a^2 + 128a^{-1}$$

Ahora se aplica la regla pasando el exponente a multiplicar y restandole 1 a dicho exponente quedando:

$$6a + 128-a^{-2}$$

$$6a - \dfrac{128}{a^2}$$





**c) Verifica la respuesta del literal anterior usando SymPy**

```python
from sympy import init_printing, symbols, diff, solve
#Definir la variable simbólica
a = symbols("a")
#Crear la función de C
C = 3*a**2+128/a
#Calcular la primera derivada e imprimirla
derivada = diff(C,a)
derivada
```

**d) Encuentra los puntos críticos de la función en forma manual. Registra el procedimiento paso a paso**

Para encontrar los puntos críticos de la función se debe igualar a 0 la derivada.

Si $6a-\dfrac{128}{a^2}=0$ entonces:

$$6a=\dfrac{128}{a^2}$$

$$6a(a^2)=128$$

$$a^3=\dfrac{128}{6}=\dfrac{64}{3}$$

$$a=\sqrt[3]{\frac{64}{3}}=\dfrac{4}{\sqrt[3]{3}}≈2.773$$

Ahora se debe reemplazar el valor de $a$ en la funcion original para conocer el punto $C(x)$

$$C\left(\dfrac{4}{\sqrt[3]{3}}\right)=3\left(\dfrac{4}{\sqrt[3]{3}}\right)^2+\frac{128}{\frac{4}{\sqrt[3]{3}}}$$

$$=3\left(\dfrac{4}{\sqrt[3]{3}}\right)^2+\dfrac{128\sqrt[3]{3}}{4}=\dfrac{48}{(\sqrt[3]{3})^2}+32\sqrt[3]{3}≈69.227$$

Con esto se obtienen los puntos críticos para $(x, y)$ estos son: $(2.773, 69.227)$






**e) Verifica la respuesta del literal anterior usando SymPy**

```python
raices = solve(derivada)

# Filtrar solo raíces reales y positivas
puntos_criticos = [
    r for r in raices
    if r.is_real and r.evalf() > 0
]

print("Los puntos críticos son:")

for pc in puntos_criticos:
    y = C.subs(a, pc)
    print(f"({pc.evalf()}, {y.evalf()})")
```

**f) Calcula la segunda derivada de la función en forma manual. Registra el procedimiento paso a paso.**

Tenemos
$$C'(a)=6a-\dfrac{128}{a^2}$$

La segunda derivada se calcula con la regla de la potencia que usamos anteriormente para calcular la primera derivada quedando.

$$6-128(-2)a^{-3}=6+\dfrac{256}{a^3}$$



**g) Verifica la respuesta del literal anterior usando SymPy**

```python
#Calcular la segunda derivada
derivada_orden_dos = diff(derivada,a)
derivada_orden_dos
```

**h) Aplicando el criterio de la segunda derivada, identifica en forma manual el mínimo de la función. Registra el procedimiento paso a paso.**

Para hallar el mínimo se debe reemplazar el valor de a por el resultado procedente de la igualación a 0 de la derivada.

$$6+\frac{256}{\frac{64}{3}}=6+\dfrac{256(3)}{64}=18$$

Este resultado es $>0$ por ende, es un mínimo

**i) Verifica la respuesta del literal anterior usando SymPy**

```python
#Mínimo de la función
for pc in puntos_criticos:

    valor_segunda = derivada_orden_dos.subs(a, pc)

    print(f"Punto crítico: {pc.evalf()}")
    print(f"Segunda derivada evaluada: {valor_segunda.evalf()}")

    if valor_segunda > 0:
        print("Es un mínimo local")

    elif valor_segunda < 0:
        print("Es un máximo local")

    else:
        print("No se puede concluir algo")
```
