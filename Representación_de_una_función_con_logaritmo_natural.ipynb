###a) Dominio y rango

**Cálculo del dominio:**
Para calcular el dominio de la función es necesario plantearla como una inecuación y hallar las incógnitas. Todo logaritmo dentro de los reales debe ser estrictamente mayor que 0, por lo cual, la ecuación se expresa de la siguiente forma:

$$\ln{(x^2-4)}>0$$

Aplicando diferencia de cuadrados queda:
$$\ln(x+2)(x-2)>0$$

$$x>-2$$ $$x<2$$

El dominio de la función entonces es:

$$(-\infty,-2)\cup(2,+\infty)$$

Esto quiere decir que la variable independiente $x$ puede tener cualquier valor $\ne-2$, $\ne2$, $\ne1$ y $\ne0$


**Verificación con SymPy:**

```python
import sympy
from sympy.calculus.util import continuous_domain
from sympy import Symbol, S

x = Symbol('x')
f = sympy.ln(x**2-4)
domain_f = continuous_domain(f, x, S.Reals)
print(f"El dominio de la función es: {domain_f}")
```

**Cálculo del rango:**

Para el rango se debe igualar la función con $f(x)$, es decir, $y$. Luego se despeja $x$.

$$y=\ln(x^2-4)$$

Se cancela el logaritmo natural con su función inversa la cual es $e$, colocándolo como base. Luego se tiene en cuenta que $e^{\ln(x)}=x$ ya que la base del logaritmo natural es $e$. Al final se realiza el despeje correspondiente.

$$e^y=e^{\ln(x^2-4)}$$

$$e^y=x^2-4$$

$$\pm\sqrt{e^y+4}=x$$

Por propiedad de las funciones exponenciales, $e^y>0$ para cualquier valor de $y \in\mathbb{R}$. Al sumar $4$ a ambos lados de esa propiedad, tenemos que $e^y+4>4$. Como el resultado siempre es mayor que 4, es evidente que $e^y + 4 \geqslant 0$.

$$(-\infty, \infty)$$




**Verificación con SymPy:**

```python
import sympy
from sympy.calculus.util import function_range, continuous_domain
from sympy import Symbol, S

x = Symbol('x')
f = sympy.ln(x**2-4)
domain_f = continuous_domain(f, x, S.Reals)

range_f = function_range(f, x, domain_f)
print(f"El rango de la función es: {range_f}")
```

###b) Puntos de corte con los ejes

**Cálculo de los puntos de corte con el eje $x$:**

Se debe igualar la función a 0 y realizar el despeje respectivo

$$0=\ln(x^2-4)$$

$$e^0=e^{\ln(x^2-4)}$$

$$1=x^2-4$$

$$5=x^2$$

$$x=\pm\sqrt{5}$$

Esto significa que la función corta al eje $x$ en dos puntos. $P(\sqrt{5},0)$ y $P(-\sqrt{5},0)$


**Verificación con SymPy:**

```python
from sympy import solve

x_intercepts = solve(f, x)
print(f"Los puntos de corte con el eje x son: {x_intercepts}")

# Para verificar los valores numéricos
for val in x_intercepts:
    print(f"Valor numérico: {val.evalf()}")
```

**Cálculo de los puntos de corte con el eje $y$:**

Se debe evaluar $x=0$ en la función para hallar el punto de corte. Sin embargo, ese valor no pertenece al dominio de la función, por ende, el punto de corte con el eje $y$ no existe.

**Verificación con SymPy:**

```python
from sympy import sympify, I

try:
    y_intercept_sym = f.subs(x, 0)
    # Comprobamos si el resultado es real o complejo
    if y_intercept_sym.is_real:
        print(f"El punto de corte con el eje y es: (0, {y_intercept_sym})")
    else:
        print("No existe un punto de corte real con el eje y.")
        print(f"Al evaluar f(0), SymPy devuelve un número complejo: (0, {y_intercept_sym})")
        print("Esto se debe a que x=0 no está en el dominio real de la función, ya que ln(0^2 - 4) = ln(-4), lo cual es un número complejo.")
except Exception as e:
    print("Hubo un error al intentar calcular el punto de corte con el eje y.")
    print(f"Razón: {e}")
    print("Esto generalmente indica que x=0 no está en el dominio de la función.")
```

###c) Asíntotas

**Cálculo de asíntotas verticales:**

Las asíntotas verticales ocurren en los logaritmos cuando el argumento se acerca a cero, ya que el logaritmo se dispara hacia ($-\infty$).Para hallarlas, se iguala el argumento del logaritmo a cero y se despeja $x$

$$x^2-4=0$$
$$x^2=4$$
$$x=\pm\sqrt{4}$$
$$x=2$$
$$x=-2$$

 Esto demuestra que a medida que la gráfica se acerca por la derecha, es decir, a $x = 2$  o por la izquierda $x = -2$, la función cae verticalmente hacia el infinito negativo sin llegar nunca a tocar esas líneas.

**Verificación con SymPy:**

```python
from sympy import Limit, oo


x = Symbol('x')
f = sympy.ln(x**2 - 4)


limit_minus_2_left = Limit(f, x, -2, dir='-').doit()
print(f"Límite cuando x tiende a -2 por la izquierda: {limit_minus_2_left}")

limit_plus_2_right = Limit(f, x, 2, dir='+').doit()
print(f"Límite cuando x tiende a 2 por la derecha: {limit_plus_2_right}")

if limit_minus_2_left == -oo and limit_plus_2_right == -oo:
    print("Se confirma que hay asíntotas verticales en x = -2 y x = 2.")
else:
    print("Las asíntotas verticales no se confirmaron con los límites calculados.")
```

**Cálculo de asíntotas horizontales:**

Para buscar asíntotas horizontales, se evalua el límite de la función cuando $x$ tiende a infinito

$$\lim_{x \to \pm\infty} \ln(x^2 - 4) = +\infty$$

Como el resultado no da un número fijo, entonces no existen asíntotas horizontales.



**Verificación con SymPy:**

```python
from sympy import Limit, oo

limit_plus_inf = Limit(f, x, oo).doit()
print(f"Límite cuando x tiende a +infinito: {limit_plus_inf}")

limit_minus_inf = Limit(f, x, -oo).doit()
print(f"Límite cuando x tiende a -infinito: {limit_minus_inf}")

if limit_plus_inf == oo and limit_minus_inf == oo:
    print("Se confirma que no existen asíntotas horizontales, ya que los límites tienden a infinito.")
else:
    print("Las asíntotas horizontales no se pudieron determinar de forma concluyente.")
```

**Cálculo de asíntotas oblicuas:**

Se debe calcular su pendiente ($m$) usando límites al infinito
$$m = \lim_{x \to \infty}\frac{f(x)}{x}$$
Aplicado a la función queda:
$$m = \lim_{x \to \infty}\frac{\ln(x^2 - 4)}{x}$$

Por regla de jerarquía de funciones se sobreentiende que la función lineal ($x$) crece muchísimo más rápido en el infinito que una función logarítmica ($\ln$). Por lo tanto, el denominador se hace infinitamente más grande que el numerador, lo que hace que el límite sea cero. Por ende
$$m = 0$$

**Verificación con SymPy:**

```python
from sympy import limit, oo, Symbol

x = Symbol('x')
f = sympy.ln(x**2 - 4)

m_limit_plus_inf = limit(f/x, x, oo)
m_limit_minus_inf = limit(f/x, x, -oo)

print(f"Límite de f(x)/x cuando x tiende a +infinito (m): {m_limit_plus_inf}")
print(f"Límite de f(x)/x cuando x tiende a -infinito (m): {m_limit_minus_inf}")

if m_limit_plus_inf == 0 and m_limit_minus_inf == 0:
    print("Como el límite de f(x)/x es 0 en ambos casos, se confirma que no existen asíntotas oblicuas.")
else:
    print("Los límites para 'm' no son finitos y distintos de cero, lo que indica que no hay asíntotas oblicuas.")
```

###d) Puntos críticos

**Cálculo de puntos críticos:**

Para calcular los puntos críticos, se halla la primera derivada de la función y luego se igualala a cero.

$$f(x) = \ln(x^2 - 4)$$

Se aplica la regla de la cadena para logaritmos, la cual dice que si $y = \ln(u)$, entonces $y' = \frac{u'}{u}$

$$f'(x) = \frac{\frac{d}{dx}(x^2 - 4)}{x^2 - 4}$$

$$f'(x) = \frac{2x}{x^2 - 4}$$

Se sobreentiende que para que 2 sea cero, x debe ser cero

$$2x = 0$$
$$x = 0$$

Además, anteriormente se calculo en las asíntotas verticales que

$$x = 2 \quad \quad x = -2$$

En esos casos la derivada no existe

**Clasificación de puntos críticos:**

Para que un valor de $x$ sea considerado un punto crítico real de la función, ese valor debe existir en la función propiamente dicha. Debe pertenecer al dominio. Esto quiere decir que la función no tiene ningún punto crítico. Esto indica que para valores menores a $-2$, la derivada es negativa, lo que significa que la función siempre es decreciente en ese trayecto. Para valores mayores a $2$, la derivada es positiva, lo que significa que la función siempre es creciente en ese trayecto.



**Verificación con SymPy:**

```python
from sympy import Symbol, diff, solve, S, Interval, Union
from sympy.calculus.util import continuous_domain

x = Symbol('x')
f = sympy.ln(x**2 - 4)

f_prime = diff(f, x)
print(f"La primera derivada f'(x) es: {f_prime}")

critical_points_from_zero = solve(f_prime, x)
print(f"Valores de x donde f'(x) = 0: {critical_points_from_zero}")

critical_points_from_undefined = solve(x**2 - 4, x)
print(f"Valores de x donde f'(x) es indefinida: {critical_points_from_undefined}")

potential_critical_points = set(critical_points_from_zero + critical_points_from_undefined)
print(f"Puntos críticos potenciales (donde f'(x) = 0 o es indefinida): {potential_critical_points}")

domain_f = continuous_domain(f, x, S.Reals)
print(f"El dominio de la función original es: {domain_f}")

actual_critical_points = []
for p in potential_critical_points:
    if domain_f.contains(p):
        actual_critical_points.append(p)

print(f"Puntos críticos reales (que pertenecen al dominio): {actual_critical_points}")

if not actual_critical_points:
    print("\nLa función no tiene puntos críticos en su dominio.")
else:
    print("\nLa función tiene puntos críticos en su dominio.")
```

###e) Crecimiento y decrecimiento

**Cálculo de intervalos de crecimiento y decrecimiento:**

Se debe construir una tabla de signos utilizando la primera derivada que calculé anteriormente. Se debe dividir la recta numérica en intervalos usando los puntos donde la derivada se hace cero ($x = 0$) o no existe ($x = -2$ y $x = 2$), pero teniendo muy en cuenta el dominio de la función original.
Intervalo 1: $(-\infty, -2)$ Intervalo 2: $(2, \infty)$

$(-\infty, -2)$

$x = -3$

$f'(-3) = \frac{2(-3)}{(-3)^2 - 4} = \frac{-6}{5}$

Es negativo. Es decir, decreciente.

$(2, \infty)$

$x = 3$
$f'(3) = \frac{2(3)}{(3)^2 - 4} = \frac{6}{5}$

Es positivo. Es decir, creciente

Esto indica que la función decrece en $(-\infty, -2)$ pero crece en $(2, \infty)$.

**Verificación con SymPy:**

```python
from sympy import Symbol, diff, S, solve, simplify, Interval, Union, solve_univariate_inequality
from sympy.calculus.util import continuous_domain

# Define the variable and function
x = Symbol('x')
f = sympy.ln(x**2 - 4)

# Get the domain of the original function
domain_f = continuous_domain(f, x, S.Reals)
print(f"El dominio de la función original es: {domain_f}")

# Calculate the first derivative
f_prime = diff(f, x)
print(f"La primera derivada f'(x) es: {f_prime}")

# Determine intervals where f'(x) > 0 (increasing) using solve_univariate_inequality
# This returns a Set object, which can be intersected
increasing_raw = solve_univariate_inequality(f_prime > 0, x, relational=False)
print(f"Intervalos donde la función es creciente (f'(x) > 0): {increasing_raw}")

# Determine intervals where f'(x) < 0 (decreasing) using solve_univariate_inequality
decreasing_raw = solve_univariate_inequality(f_prime < 0, x, relational=False)
print(f"Intervalos donde la función es decreciente (f'(x) < 0): {decreasing_raw}")

# Filter these intervals based on the function's domain
increasing_in_domain = increasing_raw.intersect(domain_f)
decreasing_in_domain = decreasing_raw.intersect(domain_f)

print(f"\nLa función es creciente en el intervalo: {increasing_in_domain}")
print(f"La función es decreciente en el intervalo: {decreasing_in_domain}")

# Compare with manual calculation results
if decreasing_in_domain == Interval.open(-S.Infinity, -2) and increasing_in_domain == Interval.open(2, S.Infinity):
    print("\nConfirmado por SymPy: Los cálculos de crecimiento y decrecimiento coinciden con el análisis manual.")
else:
    print("\nAdvertencia: Los cálculos de crecimiento y decrecimiento no coinciden con el análisis manual. Revise.")
```

###f) Curvatura

**Análisis de curvatura:**

Para la curvatura se calcula la segunda derivada de la función.

Se tiene que $f'(x) = \frac{2x}{x^2 - 4}$ Así que se aplica la regla del cociente.

$$\dfrac{2(x^2-4)-2x(2x)}{(x^2-4)^2}$$

$$\dfrac{2x^2-8-4x^2}{(x^2-4)^2}=\dfrac{-2x^2-8}{(x^2-4)^2}=\dfrac{-2(x^2+4)}{(x^2-4)^2}$$


Ahora se deben calcular los puntos de inflexión. Para ello se iguala la segunda derivada a 0, por ende se iguala el numerador a cero

$$-2(x^2 + 4) = 0$$
$$x^2 + 4 = 0$$
$$x^2 = -4$$

En los números reales, ningún número elevado al cuadrado da un resultado negativo. Eso significa que el resultado es complejo así que no se tiene en cuenta. Es decir, no existen puntos de inflexión. La función nunca cambia su tipo de curvatura.

Ahora se halla la concavidad. Para ello se analiza el signo de la segunda derivada $f''(x)$ en los intervalos del dominio.

$f''(x) = \frac{-2(x^2 + 4)}{(x^2 - 4)^2}$

El denominador $(x^2 - 4)^2$: Al estar elevado al cuadrado, su resultado siempre es positivo para cualquier valor del dominio. Además el término $(x^2 + 4)$ da siempre positivo porque cualquier número real al cuadrado es positivo o cero, y si le sumas 4, sigue siendo siempre positivo. Sumado a eso, en el numerador tenemos un $-2$ multiplicando a un término positivo. Por lo tanto, el numerador siempre es negativo. Esto significa que la función tiene una única curvatura

**Verificación con SymPy:**

```python
from sympy import Symbol, diff, solve, S, Interval, Union, solve_univariate_inequality
from sympy.calculus.util import continuous_domain

# Define the variable and function (if not already defined)
x = Symbol('x')
f = sympy.ln(x**2 - 4)

domain_f = continuous_domain(f, x, S.Reals)
print(f"El dominio de la función original es: {domain_f}")

f_prime = diff(f, x)
f_double_prime = diff(f_prime, x)
print(f"\nLa primera derivada f'(x) es: {f_prime}")
print(f"La segunda derivada f''(x) es: {f_double_prime}")

potential_inflection_points_from_zero = solve(f_double_prime, x)
print(f"Valores de x donde f''(x) = 0: {potential_inflection_points_from_zero}")

potential_inflection_points_from_undefined = solve(x**2 - 4, x)
print(f"Valores de x donde f''(x) es indefinida: {potential_inflection_points_from_undefined}")

all_potential_inflection_points = set(potential_inflection_points_from_zero + potential_inflection_points_from_undefined)
actual_inflection_points = []
for p in all_potential_inflection_points:
    if domain_f.contains(p):
        actual_inflection_points.append(p)

print(f"Puntos de inflexión reales (que pertenecen al dominio): {actual_inflection_points}")

if not actual_inflection_points:
    print("\nConfirmado: No existen puntos de inflexión para esta función en su dominio real.")
else:
    print("\nExisten puntos de inflexión en: {actual_inflection_points}")

concave_up_intervals = solve_univariate_inequality(f_double_prime > 0, x, relational=False).intersect(domain_f)
concave_down_intervals = solve_univariate_inequality(f_double_prime < 0, x, relational=False).intersect(domain_f)

print(f"\nLa función es cóncava hacia arriba en los intervalos: {concave_up_intervals}")
print(f"La función es cóncava hacia abajo en los intervalos: {concave_down_intervals}")

if not concave_up_intervals and concave_down_intervals == domain_f:
    print("La función es siempre cóncava hacia abajo en todo su dominio, coincidiendo con el análisis manual.")
elif not concave_up_intervals and not concave_down_intervals:
    print("\nAdvertencia: SymPy no pudo determinar claramente la concavidad. Revise.")
else:
    print("\nAdvertencia: Los cálculos de concavidad no coinciden con el análisis manual. Revise.")
```

###g) Gráfica

```python
import numpy as np
import matplotlib.pyplot as plt

x_izq = np.linspace(-6, -2.001, 400)
x_der = np.linspace(2.001, 6, 400)

y_izq = np.log(x_izq**2 - 4)
y_der = np.log(x_der**2 - 4)

plt.figure(figsize=(10, 6))

plt.plot(x_izq, y_izq, color='blue', linewidth=2, label=r'$f(x) = \ln(x^2 - 4)$')
plt.plot(x_der, y_der, color='blue', linewidth=2)

plt.axvline(x=-2, color='red', linestyle='--', linewidth=1.5, label=r'Asíntotas verticales ($x = \pm 2$)')
plt.axvline(x=2, color='red', linestyle='--', linewidth=1.5)

raiz_1 = np.sqrt(5)
raiz_2 = -np.sqrt(5)
plt.scatter([raiz_2, raiz_1], [0, 0], color='black', zorder=5)
plt.text(raiz_1 + 0.1, 0.3, r'$(\sqrt{5}, 0)$', fontsize=10, fontweight='bold')
plt.text(raiz_2 - 0.7, 0.3, r'$(-\sqrt{5}, 0)$', fontsize=10, fontweight='bold')

plt.axhline(0, color='black', linewidth=1) # Eje X
plt.axvline(0, color='black', linewidth=1) # Eje Y
plt.grid(True, which='both', linestyle=':', alpha=0.6)

plt.title(r'Análisis Gráfico de $f(x) = \ln(x^2 - 4)$', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Eje X', fontsize=12)
plt.ylabel('Eje Y', fontsize=12)
plt.xlim(-6, 6)
plt.ylim(-4, 4)
plt.legend(loc='upper center', fontsize=10)

plt.show()
```
