###a) Dominio y rango

**Cálculo del dominio:**
Para calcular el dominio de la función es necesario plantearla como una inecuación y hallar las incógnitas. Todo logaritmo dentro de los reales debe ser estrictamente mayor que 0, por lo cual, la ecuación se expresa de la siguiente forma:

$$\ln{(x^2-4)}>0$$

Aplicando diferencia de cuadrados queda:
$$\ln(x+2)(x-2)>0$$

$$x>-2$$ $$x<2$$

El dominio de la función entonces es:

$$(-\infty,-2)\cup(2,+\infty)$$

Esto quiere decir que la variable independiente $x$ puede tener cualquier valor $\ne-2$, $\ne2$ y $\ne0$


**Verificación con SymPy:**

```python
import sympy
from sympy.calculus.util import continuous_domain
from sympy import Symbol, S

x = Symbol('x')
f = sympy.ln(x**2-4)
domain_f = continuous_domain(f, x, S.Reals)
print(f"El dominio de la función es: {domain_f}")
```

**Cálculo del rango:**

Para el rango se debe igualar la función con $f(x)$, es decir, $y$. Luego se despeja $x$.

$$y=\ln(x^2-4)$$

Se cancela el logaritmo natural con su función inversa la cual es $e$, colocándolo como base. Luego se tiene en cuenta que $e^{\ln(x)}=x$ ya que la base del logaritmo natural es $e$. Al final se realiza el despeje correspondiente.

$$e^y=e^{\ln(x^2-4)}$$

$$e^y=x^2-4$$

$$\pm\sqrt{e^y+4}=x$$

Por propiedad de las funciones exponenciales, $e^y>0$ para cualquier valor de $y \in\mathbb{R}$. Al sumar $4$ a ambos lados de esa propiedad, tenemos que $e^y+4>4$. Como el resultado siempre es mayor que 4, es evidente que $e^y + 4 \geqslant 0$.

$$(-\infty, \infty)$$




**Verificación con SymPy:**

```python
import sympy
from sympy.calculus.util import function_range, continuous_domain
from sympy import Symbol, S

x = Symbol('x')
f = sympy.ln(x**2-4)
domain_f = continuous_domain(f, x, S.Reals)

range_f = function_range(f, x, domain_f)
print(f"El rango de la función es: {range_f}")
```

###b) Puntos de corte con los ejes

**Cálculo de los puntos de corte con el eje $x$:**

Se debe igualar la función a 0 y realizar el despeje respectivo

$$0=\ln(x^2-4)$$

$$e^0=e^{\ln(x^2-4)}$$

$$1=x^2-4$$

$$5=x^2$$

$$x=\pm\sqrt{5}$$

Esto significa que la función corta al eje $x$ en dos puntos. $P(\sqrt{5},0)$ y $P(-\sqrt{5},0)$


**Verificación con SymPy:**

```python
from sympy import solve

x_intercepts = solve(f, x)
print(f"Los puntos de corte con el eje x son: {x_intercepts}")

# Para verificar los valores numéricos
for val in x_intercepts:
    print(f"Valor numérico: {val.evalf()}")
```

**Cálculo de los puntos de corte con el eje $y$:**

Se debe evaluar $x=0$ en la función para hallar el punto de corte. Sin embargo, ese valor no pertenece al dominio de la función, por ende, el punto de corte con el eje $y$ no existe.

**Verificación con SymPy:**

```python
from sympy import sympify, I

try:
    y_intercept_sym = f.subs(x, 0)
    # Comprobamos si el resultado es real o complejo
    if y_intercept_sym.is_real:
        print(f"El punto de corte con el eje y es: (0, {y_intercept_sym})")
    else:
        print("No existe un punto de corte real con el eje y.")
        print(f"Al evaluar f(0), SymPy devuelve un número complejo: (0, {y_intercept_sym})")
        print("Esto se debe a que x=0 no está en el dominio real de la función, ya que ln(0^2 - 4) = ln(-4), lo cual es un número complejo.")
except Exception as e:
    print("Hubo un error al intentar calcular el punto de corte con el eje y.")
    print(f"Razón: {e}")
    print("Esto generalmente indica que x=0 no está en el dominio de la función.")
```
